# V2 itinerary maps

This notebook maps the original itinerary from `itinerary.json` and the optimized multi-day VRP result from `vrp_results.json`.

```python
# !pip install folium
```

```python
from pathlib import Path

from map_itineraries import (
    build_maps,
    create_folium_map,
    hotel_from_vrp,
    latest_vrp_result,
    load_json,
    optimized_vrp_routes,
    original_itinerary_routes,
)

base_dir = Path.cwd()
if not (base_dir / "itinerary.json").exists():
    base_dir = Path("optimization/V2")

original_path = base_dir / "itinerary.json"
optimized_path = base_dir / "vrp_results.json"

original_itinerary = load_json(original_path)
vrp_result = latest_vrp_result(load_json(optimized_path))
hotel_location = hotel_from_vrp(vrp_result)

hotel_location
```

```python
print("Generating original itinerary map...")
original_map = create_folium_map(
    original_itinerary_routes(original_itinerary),
    title="Original itinerary",
    hotel_location=hotel_location,
)
original_map
```

```python
print("Generating optimized VRP itinerary map...")
optimized_map = create_folium_map(
    optimized_vrp_routes(vrp_result),
    title="Optimized VRP itinerary",
    hotel_location=hotel_location,
)
optimized_map
```

```python
original_map, optimized_map = build_maps(
    original_path=original_path,
    optimized_path=optimized_path,
    output_dir=base_dir,
)

print("Saved original_itinerary_map.html")
print("Saved optimized_vrp_itinerary_map.html")
```
